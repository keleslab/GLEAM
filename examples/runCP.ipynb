```python
import numpy as np
import pandas as pd
from gleam.CP import run_gleam_conformal

# 1. Create Dummy Data
n_samples = 100
n_genes = 10

# Simulated Predictions
X_pred = np.random.normal(10, 2, size=(n_samples, n_genes))
P_zero = np.random.uniform(0, 0.5, size=(n_samples, n_genes)) # Prob of zero

# Simulated Ground Truth (inject zeros)
X_true = X_pred.copy()
mask = np.random.choice([True, False], size=X_true.shape, p=[0.2, 0.8])
X_true[mask] = 0

# 2. Split into Test and Calibration (50/50)
idx = int(n_samples / 2)
test_pred, calib_pred = X_pred[:idx], X_pred[idx:]
test_prob, calib_prob = P_zero[:idx], P_zero[idx:]
test_true, calib_true = X_true[:idx], X_true[idx:]

# 3. Run Conformal Prediction
results = run_gleam_conformal(
    pred_val=test_pred, 
    pred_prob=test_prob, 
    true_val=test_true,
    calib_pred_val=calib_pred,
    calib_pred_prob=calib_prob,
    calib_true_val=calib_true,
    alpha=0.1  # 90% Coverage target
)

# 4. Inspect Results
print(f"Classification Coverage: {results['classification']['coverage']:.4f}")

df_reg = results['regression']
if df_reg is not None:
    print("\nRegression Intervals (First 5 Non-Zeros):")
    print(df_reg.head())
    
    # Check Regression Coverage manually
    in_interval = (df_reg['true'] >= df_reg['lower']) & (df_reg['true'] <= df_reg['upper'])
    print(f"\nRegression Coverage: {in_interval.mean():.4f}")
```
